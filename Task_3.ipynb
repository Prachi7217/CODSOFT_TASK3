```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

data = {
    "Product": [
        "Laptop", "Phone", "Headphones", "Laptop", "Phone",
        "Tablet", "Laptop", "Headphones", "Phone", "Tablet",
        "Laptop", "Phone", "Headphones", "Tablet", "Laptop"
    ],
    "Region": [
        "West", "North", "West", "South", "North",
        "East", "West", "South", "East", "North",
        "South", "West", "North", "East", "West"
    ],
    "Quantity": [2, 5, 10, 1, 4, 3, 2, 8, 6, 2, 3, 5, 12, 4, 2],
    "Sales": [
        100000, 75000, 10000, 50000, 60000,
        45000, 100000, 8000, 90000, 30000,
        150000, 75000, 12000, 60000, 100000
    ]
}

df = pd.DataFrame(data)

print("Dataset loaded successfully!")
df.head()
```

```python
product_sales = df.groupby("Product")["Sales"].sum()

plt.figure(figsize=(8, 5))
product_sales.plot(kind="bar", color="skyblue")

plt.title("Total Sales by Product")
plt.xlabel("Product")
plt.ylabel("Total Sales (₹)")
plt.xticks(rotation=0)
plt.tight_layout()
plt.show()
```

```python
plt.figure(figsize=(9, 5))
plt.plot(
    range(1, len(df) + 1),
    df["Sales"],
    marker="o",
    color="green",
    label="Sales"
)

plt.title("Sales by Transaction")
plt.xlabel("Transaction Number")
plt.ylabel("Sales (₹)")
plt.legend()
plt.grid(True)
plt.tight_layout()
plt.show()
```

```python
plt.figure(figsize=(7, 7))

product_sales.plot(
    kind="pie",
    autopct="%1.1f%%",
    startangle=90
)

plt.title("Product-wise Sales Share")
plt.ylabel("")
plt.tight_layout()
plt.show()
```

```python
plt.figure(figsize=(8, 5))

plt.hist(df["Sales"], bins=6, color="orange", edgecolor="black")

plt.title("Distribution of Sales")
plt.xlabel("Sales Amount (₹)")
plt.ylabel("Frequency")
plt.tight_layout()
plt.show()
```

```python
plt.figure(figsize=(8, 5))

sns.scatterplot(
    data=df,
    x="Quantity",
    y="Sales",
    hue="Region",
    s=100
)

plt.title("Quantity vs Sales by Region")
plt.xlabel("Quantity Sold")
plt.ylabel("Sales (₹)")
plt.legend(title="Region")
plt.tight_layout()
plt.show()
```

```python
fig, axes = plt.subplots(2, 2, figsize=(14, 10))

# Bar chart
product_sales.plot(kind="bar", ax=axes[0, 0])
axes[0, 0].set_title("Sales by Product")
axes[0, 0].set_xlabel("Product")
axes[0, 0].set_ylabel("Sales (₹)")

# Line chart
axes[0, 1].plot(range(1, len(df) + 1), df["Sales"], marker="o")
axes[0, 1].set_title("Sales by Transaction")
axes[0, 1].set_xlabel("Transaction Number")
axes[0, 1].set_ylabel("Sales (₹)")

# Histogram
axes[1, 0].hist(df["Sales"], bins=6, edgecolor="black")
axes[1, 0].set_title("Sales Distribution")
axes[1, 0].set_xlabel("Sales (₹)")
axes[1, 0].set_ylabel("Frequency")

# Scatter plot
sns.scatterplot(
    data=df, x="Quantity", y="Sales",
    hue="Region", ax=axes[1, 1]
)
axes[1, 1].set_title("Quantity vs Sales")

plt.tight_layout()
plt.show()
```
